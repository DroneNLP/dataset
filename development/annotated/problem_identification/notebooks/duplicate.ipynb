```python
# Duplicate
import pandas as pd

label_map = {
    'normal': 'Normal',
    'low': 'Low',
    'medium': 'Medium',
    'high': 'High'
}

unfiltered_test = pd.read_csv('severity/duplicate/unfiltered_test.csv')
unfiltered_test['label'] = unfiltered_test['label'].map(label_map)
unfiltered_test.to_csv('severity/duplicate/unfiltered_test.csv', index=False)
```

```python
unfiltered_train = pd.read_csv('severity/duplicate/unfiltered_train.csv')
unfiltered_train['label'] = unfiltered_train['label'].map(label_map)
unfiltered_train.to_csv('severity/duplicate/unfiltered_train.csv', index=False)
```
